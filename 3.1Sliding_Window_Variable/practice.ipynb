```python
def lengthofLongestSubstring(s):
    window = set()
    left = 0
    max_length = 0

    for right in s:
        while right in window:
            window.remove(s[left])
            left += 1
        window.add(right)

        max_length = max(max_length,len(window))

    return max_length

#s = "abcabcabcd"
s = "pwwkew"
print(lengthofLongestSubstring(s))
```

```python
'''
Fruit into Basket

Example 1:

Input: fruits = [1,2,1]
Output: 3
Explanation: We can pick from all 3 trees.
Example 2:

Input: fruits = [0,1,2,2]
Output: 3
Explanation: We can pick from trees [1,2,2].
If we had started at the first tree, we would only pick from trees [0,1].
Example 3:

Input: fruits = [1,2,3,2,2]
Output: 4
Explanation: We can pick from trees [2,3,2,2].
If we had started at the first tree, we would only pick from trees [1,2].

'''
```

```python
from collections import defaultdict
def totalFruits(fruits):
    seen = defaultdict(int)
    left = 0
    max_len = 0

    for right in range(len(fruits)):
        seen[fruits[right]] += 1

        while len(seen) > 2:
            seen[fruits[left]] -= 1

            if seen[fruits[left]] == 0:
                seen.pop(fruits[left])
            left += 1
        max_len = max(max_len, right - left + 1)

    return max_len

fruits = [1,2,3,2,2,3,3,2,1,1,2,3,3,2,2,1,1,3]

print(totalFruits(fruits))

                 

```

```python
'''
Longest repeating Character Replacement

Example 1:

Input: s = "ABAB", k = 2
Output: 4
Explanation: Replace the two 'A's with two 'B's or vice versa.
Example 2:

Input: s = "AABABBA", k = 1
Output: 4
Explanation: Replace the one 'A' in the middle with 'B' and form "AABBBBA".
The substring "BBBB" has the longest repeating letters, which is 4.
There may exists other ways to achieve this answer too.

'''
```

```python
def characterReplacement(s, k):
    freq_map = {}
    left = 0
    max_freq = 0

    for right in range(len(s)):
        freq_map[s[right]]  = 1 + freq_map.get(s[right], 0)
        max_freq = max(max_freq, freq_map[s[right]])

        window_len = right - left + 1
        replacement = window_len - max_freq

        if replacement > k:
            freq_map[s[left]] -= 1
            left += 1

    return right - left + 1


s = "ABAB"
k = 2

print(characterReplacement(s,k))
```

```python
#s = "pwwkew"
s = "abcabcabcd"

window = set()
max_substring = 0
left = 0

for right in s:
    print(right)
    if right not in window:
        window.add(right)
    else:
        print(f"Current Window: {window}")
        current_substring = len(window)
        if current_substring > max_substring:
            max_substring = current_substring

        while right in window:
            window.remove(s[left])
            left += 1
        window.add(right)

if len(window) > max_substring:
    max_substring = len(window)

print(max_substring)
```
